```python
import time, json, struct
import queue
from pathlib import Path

# Fixed import - use FrameReader, not SerialFrameReader
from interrogator_stream import FrameReader

def record_chunk(port="COM15", baud=921600, duration_s=10.0,
                 raw_bin="chunk.bin", raw_jsonl="chunk_raw.jsonl"):
    q = queue.Queue(maxsize=5000)
    reader = FrameReader(port, baud, q)  # No extra raw_bin/raw_jsonl params needed
    reader.start()

    t0 = time.time()
    n = 0
    try:
        while time.time() - t0 < duration_s:
            try:
                _ = q.get(timeout=0.2)
                n += 1
            except queue.Empty:
                pass
    finally:
        reader.stop()
        reader.join(timeout=1.0)  # Added join for clean shutdown

    print(f"Recorded {n} frames")

record_chunk(duration_s=5.0)
```
